```kotlin
%use kandy
%use dataframe
%use lets-plot
```

```kotlin
val url = "http://192.168.35.107:7788/khnp/wastewater"
val df = DataFrame.readJson(url)
df.head(5)
```

```kotlin
import kotlin.math.round

val convertValueDf = df.convert { tm001 and tm002 }.with {
    val value = it?.toString()?.toDoubleOrNull() ?: 0.0
    round(value * 10) / 10.0
}
convertValueDf.head(5)
```

```kotlin
import kotlinx.datetime.*
import kotlin.time.Duration.Companion.minutes

val convertTimeDf = convertValueDf.convert {
    time and tm001_time and tm002_time
}.with { 
    LocalDateTime.parse(it.toString().replace(" ", "T")).toInstant(TimeZone.UTC)
}

convertTimeDf.head(5)
```

```kotlin
val checkTime = 10.minutes

val finalDf = convertTimeDf.filter {
    (time - tm001_time).absoluteValue <= checkTime &&
    (time - tm002_time).absoluteValue <= checkTime 
}
finalDf.head(5)
```

```kotlin
//letsPlot( finalDf.filter{ time >= "2026-04-21 22:00" }.toMap() ) +

val chart = letsPlot( finalDf.toMap() ) +
labs( title = "분단위 실시간 폐수 수질현황", y="Wastewater Quality (PH)", x="수집시간", color="발전소") +     
geomPoint( alpha = 0.6){
    x = "time";
    y = "tm002";
    color="genName";
    size="tm001";
} +
scaleYContinuous(limits=Pair(5.0,9.0)) +
// %Y: 년, %m: 월, %d: 일, %H: 시, %M: 분
scaleXDateTime(
    format = "%m-%d %H:%M" 
) +
scaleSize(range=Pair(0, 15), name="유량") +
theme(
    axisTextX= elementText( angle=45, hjust = 1.0),
) +    
ggsize( width = 2400, height = 450)    
```

```kotlin
import org.jetbrains.letsPlot.export.ggsave

ggsave(
    plot = chart, 
    filename = "chart.svg", 
)
```

```kotlin
val url = "http://192.168.35.107:7788/khnp/thermalwastewater"
val df = DataFrame.readJson(url)
df.head(5)
```

```kotlin
import kotlinx.datetime.*
import kotlin.time.Duration.Companion.minutes

val convertedDf = df.convert {
    time and rm001_time and rm002_time and rm005_time and rm006_time
}.with { 
    LocalDateTime.parse(it.toString().replace(" ", "T")).toInstant(TimeZone.UTC)
}

convertedDf.head(5)
```

```kotlin
val checkTime = 20.minutes

val finalDf = convertedDf.filter {
    (time - rm001_time).absoluteValue <= checkTime &&
    (time - rm002_time).absoluteValue <= checkTime &&
    (time - rm005_time).absoluteValue <= checkTime &&
    (time - rm006_time).absoluteValue <= checkTime
}
finalDf.head(5)
```

```kotlin
val data = finalDf.select{ time and genName and rm001 and rm005}
data.describe()
```

```kotlin
import kotlin.math.round

val selectDf = finalDf.select{ time and genName and rm001 and rm005}
    .convert{ rm001 and rm005}.with {
        val value = it?.toString()?.toDoubleOrNull() ?: 0.0
        round(value * 10) / 10.0
}
selectDf.describe()
```

```kotlin
val data = selectDf.toMap()
data
```

```kotlin

val chart2 = letsPlot(data) +
geomRibbon(
    alpha = 0.2,

    tooltips = layerTooltips()
        .title("@genName")              // 툴팁 제목에 발전소 이름 표시
        .line("시간|@time")             // '라벨|값' 형식으로 표시
        .line("취수 수온|@rm001 °C")
        .line("배수 수온|@rm005 °C")
        .format("@time", "%m-%d %H:%M") // 시간 형식 지정
        .format("@rm001", ".1f")        // 소수점 첫째 자리까지 표시
        .format("@rm005", ".1f")
          ){
    x="time";
    ymin = "rm001";
    ymax = "rm005";
    fill = "genName";
} +
scaleYContinuous(limits=Pair(5.0, 30.0)) +    
scaleXDateTime(  
    format = "%m-%d %H:%M" ,
 //   breaks = data["time"] as List<*> 
) +
labs( title = "발전소별 실시간 취수/배수 수온 현황", y="수온(°C)", x="수집시간", fill="발전소") +  
theme(
    axisTextX= elementText( angle=45),
    panelGridMajorX = elementLine(color = "#e0e0e0", size = 0.5),
) +
ggsize( width = 2400, height = 450)   



```

```kotlin
ggsave(
    plot = chart2, 
    filename = "chart2.svg", 
)
```
