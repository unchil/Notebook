```kotlin
fun getDummies(colunmName: DataColumn<String>):DataFrame<Any?>{
    return colunmName.map {
            val list = mutableListOf<Boolean>()
            colunmName.distinct().forEach {item ->
                list.add( it.equals(item))
            }
            list
        }.toDataFrame()
}
```

```kotlin
val specieDummies = getDummies(iris.species)
```

```kotlin
val irisDf = iris.add(
    specieDummies.split{ species }.into { "specie_$it" }
)
irisDf.head()
```

```kotlin
val data =  irisDf.select { sepal_length and sepal_width and petal_length and petal_width }
    .map {
        it.values()
    }
```
