```kotlin
import org.json.XML
```

```kotlin
val serviceKeyFilePath = "/Volumes/WorkSpace/Notebook/data/OceanMensurationService.json"
```

```kotlin
val serviceInfo = DataRow.readJson(path=serviceKeyFilePath)
```

```kotlin
val codeUrl = "${serviceInfo.endpoint_json}?id=${serviceInfo.code}&key=${serviceInfo.key}"
```

```kotlin
import io.ktor.client.request.*
import io.ktor.http.*
import io.ktor.utils.io.charsets.*

val codeResponse = http.get(codeUrl){
  //  contentType(ContentType.Application.Json.withCharset(Charset.forName("EUC-KR")) )
}
codeResponse
```

```kotlin
val codeRawDf = DataFrame.readJsonStr( codeResponse.bodyAsText(Charset.forName("EUC-KR")) )
codeRawDf
```

```kotlin
val codeDf = codeRawDf.body.item.concat()
codeDf
```

```kotlin
val listUrl = "${serviceInfo.endpoint_json}?id=${serviceInfo.list}&key=${serviceInfo.key}"
val listRawDf = DataFrame.readJsonStr( http.get(listUrl).bodyAsText(Charset.forName("EUC-KR")) )
listRawDf
```

```kotlin
val listDf = listRawDf.body.item.concat()
listDf
```

```kotlin
val df = listDf.innerJoinWith(codeDf) {
    right.sta_cde == sta_cde
}
df
```

```kotlin
df.filter { gru_nam.equals("동해") && repair_gbn.equals("1") }
    .select { sta_nam_kor and obs_lay and wtr_tmp }
    .convert{ wtr_tmp }.with { it.toFloat() }
    .sortBy { sta_nam_kor and obs_lay  }
    .groupBy("obs_lay")
    .plot{
        layout{
            title = "Sea Water Quality"
            size = 1000 to 400
        }
        bars{
            x("sta_nam_kor"){
                axis{
                    name = "관측지점"
                }
            }
            y("wtr_tmp"){
                scale = continuous(0.0..15.0)
                axis {
                    name = "온도 °C"
                }
            }
            fillColor("obs_lay"){
                scale = categorical(
                    listOf(Color.BLUE, Color.GREEN, Color.RED),
                    listOf( "3", "2", "1"),
                )
                legend{
                    name= "관측 수심"
                    breaksLabeled("1" to "표층", "2" to "중층", "3" to "저층")
                }
            }
        }
    }
```
